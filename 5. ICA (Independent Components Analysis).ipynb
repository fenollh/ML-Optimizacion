## **5. ICA (Independent Components Analysis)**

En este cuaderno introduciremos otro algoritmo de aprendizaje no supervisado pero diseñador para resolver otro tipo de problemas.
Supongamos que tenemos un dataset $\{x^{(i)}\}_i$ donde las observaciones asumimos que provienen de una fuentes las cuales se han mezclado y nuestro objetivo es usar la informacion contenida en el dataset para aislar y recuperar cada fuente particular. Mas concretamente, si llamamos $s$ a las fuentes deonde el superindice indicara instante temporal y el subindice indicara a cual de las $d$ fuentes nos referimos, entones suponemos:
$$
x = As
$$
Donde A es la matriz que mezcla las fuentes. Por lo tanto el objetivo principal será encontrar $W=A^{-1}$ de forma que a partir de los datos podamos recuperar las fuentes aisladas $s=Wx$. Esta matriz W serán los parámetros que trataremos de encontrar. \
En primer lugar llamaremos $p_j$ a la funcion de densidad de la fuente $s_j$. De esta forma, asumiendo que son independientes la funcion de densidad conjunta será:
$$
p_s(s) = \prod p_j(s_j)
$$
Como $s = Wx$ tenemos que: 
$$
p(x) = \prod p_j(w_j^Tx_j)|W|
$$
Ahora lo que haremos sera especificar distribuciones marginales para las fuentes, lo haremos especificando una funcion de distribucion que en todo caso asumiremos que es la sigmoide y la llamaremos $g$ de forma que escribimos $p_j(\cdot) := g'(\cdot)$. En caso de tener conocimiento experto sobre las distribuciones de las fuentes podriamos usar otra distribuion que se adaptase a ese conocimiento, el usar la sigmoide es porque, en caso de que no dispongamos de ese conocimiento, suele funcionar correctamente. De esta forma tenemos que la densidad de cada ejemplo $x$ es de la forma:
$$
p(x) = \prod g'(w_j^Tx_j^{(i)})|W|
$$
Los parametros $W$ los seleccionaremos maximizando la log-verosimilitud de la muestra, es decir maximizando:
$$
\ell(W) = \sum_i^n\sum_j^d \log g'(w_j^Tx_j^{(i)})+\log|W|
$$
respecto de $W$. Este se hará mediante ascendo de gradiente estocastico. Si los datos no eran independientes, por ejemplo en el caso de estra tratando series temporales, se usara un dataset mezclado. Realizando los calculos la regla de actualizacion del gradiente queda como:
$$
W \leftarrow W + \alpha
\begin{bmatrix}
\frac{g''(w_1^Tx_1^{(i)})}{g'(w_1^Tx_1^{(i)})}(x^{(i)})^T\\
\frac{g''(w_2^Tx_2^{(i)})}{g'(w_2^Tx_2^{(i)})}(x^{(i)})^T\\
\vdots\\
\frac{g''(w_d^Tx_d^{(i)})}{g'(w_d^Tx_d^{(i)})}(x^{(i)})^T\\
\end{bmatrix}
+\alpha (W^T)^{-1}
$$
