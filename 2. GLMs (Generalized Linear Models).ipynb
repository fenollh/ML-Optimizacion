# **Generalized Linear Models (GLM)**

En este capitulo desarrollaremos el tema de los modelos lineales generalizados y la implmentacion de algunos de ellos como regresion lineal, regresion logistica, softmax o regresion lineal localmente ponderada.

## La familia exponencial
En primer lugar debere recordar ciertos conceptos estudiandos en la asignatura de Estadistica. En particular los GLM se obtendran a partir de la llamada familia exponencial.  
Se dice que una familia de distribuciones de probabilidad pertenece a la familia exponencial si su funcion de densidad se puede escribir de la forma:
$$
p(y;\eta) = b(y)\exp\!\left( \eta^{T} T(y) - a(\eta)\right)
$$
donde:  
$\eta$ es el parametro natural  
$b(y)$ es la funcion base que solo depende de los datos  
$T(y)$ es un estadistico suficiente de y  
$a(\eta)$ es la funcion log-particion que solo depende del parametro natural  
Las distribuciones que pertenecen a esta familia tiene propiedades interesantes como que la funcion de verosimilitud existe y es convexa que la media y la varianza derivan de a() de la forma $\mathbb{E}[T(y)] = a'(\eta)$.  
Ejemplos de distribuciones que pertenecen a la familia exponencial son la Bernouilli($\eta$), la Poisson($\eta$), la Normal($\eta$), la Gamma($\eta$), ..., notese que $\eta$ puede ser un escalar como en la Bernouilli o un vector si consideramos la Normal con varianza y media desconocidas.


## Conexion con **Regresion Lineal**
Ahora volvamos al caso de regresion lineal. En este caso asumimos que nuestros datos se pueden modelan como $y=\theta x+\varepsilon$ donde el epsilon es el ruido y es de la forma $\varepsilon \sim \mathcal{N}(0,\sigma^{2})$. Entoces estamos asumiendo que $y|x \sim \mathcal{N}(\theta x,\sigma^{2})$.  
  
  Por lo tanto estamos buscando una funcion $h_\theta (x)$ que me devuelva $\mathbb{E}[y|x]$    
  
Pero bajo nuestra asunciones $y|x \sim \mathcal{N}(\theta x,\sigma^{2})$ y sabemos que la normal es una distribucion que pertenece a la familia exponencial. En general la densidad de una normal (mu, 1) se puede escribir como de la familia exponencia con:    

$T(y) = y$  
$\eta = \mu$  
$a(\eta) = \frac{\mu^2}{2}$    

De donde llegamos a $\mathbb{E}[y|x] = a'(\eta) = \eta = \theta x := h_\theta (x)$ de donde hemos definido el modelo de regresion lineal. Ahora falta escoger theta, lo haremos mediante la funcion de verosimilitud: 
$$
L(\theta; y_1, \dots, y_n \mid x_1, \dots, x_n) 
= \prod_{i=1}^n \frac{1}{\sqrt{2\pi}} 
\exp\!\left( -\frac{1}{2} \big(y_i - \theta x_i \big)^2 \right).
$$  
que usando que el logaritmo es monotona cerciente e inyectiva tenemos que maximizar la verorimilitud es equivalente a maximizar la log-verosimilitud: 
$$
\ell(\theta) 
= -\frac{n}{2} \log(2\pi) 
- \frac{1}{2} \sum_{i=1}^n \big(y_i - \theta x_i \big)^2.
$$
y aqui tenemos que $-\frac{n}{2} \log(2\pi)$ es una constante y por tanto el objetivo sera maximizar $-\frac{1}{2} \sum_{i=1}^n \big(y_i - \theta x_i \big)^2$ que es igual a minimizar $\frac{1}{2} \sum_{i=1}^n \big(y_i - \theta x_i \big)^2$ y esto es la funcion de perdida que definimos en el algoritmo de regresion lineal formulado de manera estandar, los minimos  cuadrados.
  
**De esta manera hemos justificado de manera teorica como si asumimos que nuestra poblacion se distribuye siguiendo una normal y la relacion entre los y y los x es lineal entonces la eleccion de  $h_\theta(x) := \theta x$ esta justificada al ser el valor esperado de y dado x y la eleccion de $\frac{1}{2} \sum_{i=1}^n \big(y_i - \theta x_i \big)^2$ como funcion de perdida tambien al ser lo que hay que minimizar para obtener la maxima verosimilitud de los parametros con los datos que tenemos.**  
Ahora veamos que ocurre si asumimos que la variable y no se distribuye como una normal. Si no se distribuye como una normal pero si como otra distribucion que pertenezca a la familia exponencial entonces podemos seguir un procedimiento analogo, veamos el caso en que se distribuye como una $Ber(\eta)$


## **Regresion Logistica**
En el apartado anterior asumiamos que $y|x \sim \mathcal{N}(\eta,\sigma^{2})$. Ahora supongamos que tenemos un problema en que los unicos valores de y son 0 o 1. En este caso nuestra variable y ya no se distribuye como una normal si no mas bien como una Bernouilli, que tambien pertenece a la familia exponencial. Ahora como estamos diseñando un modelo lineal imponemos que el parametro $\eta$ de la distribucion es $\theta x$ y hagamos un desarrollo parecido al de antes.  
La $Ber(p)$ pertenece a la familia exponencial haciendo:  

$T(y) = y$  
$\eta = log(\frac{p}{1-p})  \qquad (p=\frac{1}{1+e^{-\eta}})$  
$a(\eta) = log(1+e^\eta)$  

Ahora por la propiedades vistas antes tenemos: $\mathbb{E}[y|x] = a'(\eta) = \frac{e^\eta}{1+e^\eta} = \frac{1}{1+e^{-\eta}}  = \frac{1}{1+e^{-\theta x}}:= h_\theta (x)$ que sera nuestra funcion hipotesis en el modelo de regresion logistica.  
Ahora para seleccionar el theta optimo procederemos como antes. En este caso la verosimilitud de theta sera:
$$
L(\theta; y_1, \dots, y_n \mid x_1, \dots, x_n) = \prod_{i=1}^{n} h_\theta(x_i)^{y_i} \, \big( 1 - h_\theta(x_i) \big)^{1 - y_i}
$$
y la log-verosimilitud:  
$$
\ell(\theta) = \sum_{i=1}^{n} \Big( y_i \log h_\theta(x_i) + (1 - y_i) \log (1 - h_\theta(x_i)) \Big)
$$
que tiene gradiente:
$$
\nabla_\theta \ell(\theta) = \sum_{i=1}^{n} x_i \, \big( y_i - h_\theta(x_i) \big)
$$
Ahora con estos datos se aplica un metodo de optimizacion para maximizar la verosimilitud de los parametros.  
Implementemos estos dos algoritmos que hemos derivado hasta ahora:

## Apendice. Optimizadores
Vamos a implementar algunos optimizadores, en concreto gradiente descend y Newton para optimizar los parametros, No hace falta prestar demasiada atencion a esta seccion, la mayoria son copia y pega de la anterior.

```python
def gradient_descent (f, Df, x0, alfa=0.1, itmax=200, tol=1e-3):
    x = np.array(x0, dtype=float).copy() #partimos del dato inical

    for i in range(itmax):
        Dfx = Df(x)

        #criterio de parada
        if np.linalg.norm(Dfx)<tol:
            break
        x_next = x - alfa*Dfx
        x = x_next
    return x


def newton (f, Df, x0, itmax=200, tol=1e-3):
    #Para aplicar Newton, como vamos a tener que invertir la hessiana, necesitamos que esta sea una tensor de orden <=2 
    #y por lo tanto debemos aplanar todos nuestros datos antes de hacer los calculos y volverlos a reshapear al hacer el return 
    m=x0.shape[0]
    k=x0.shape[1]
    x = np.array(x0, dtype=float).copy().flatten() #partimos del dato inical
    f_flat = lambda theta: f(theta.reshape(m, k))
    Hf = hessian(f_flat)
    Df = grad(f_flat)
    
    for i in range(itmax):
        Hfx = Hf(x)
        Dfx = Df(x)
        
        #criterio de parada
        if np.linalg.norm(Dfx)<tol:
            break

        delta_x = -anp.linalg.solve(Hfx, Dfx) #Es equivalente a H-1.D pero mas eficiente
        x = x + delta_x
        
    return x.reshape(m, k)
```

## **Implementacion**

```python
import numpy as np
import autograd.numpy as anp
from autograd import grad, hessian
import matplotlib.pyplot as plt
```

```python
def linear_regression(X, y, itmax=200, optimizer=gradient_descent):
    '''
    Para este caso tenemos que hay que estimar 1 parametro por variable a partir de un dataset con n ejemplos con m variables cada uno. 
    Por lo tanto las dimensiones son:
    X (Dataset):  nxm
    y (Labels): Como entrada 1xn
    theta (Parametros): 1xm
    '''
    #Definimos la funcion hipotesis
    h = lambda theta, x: x @ theta
    
    #Definimos la funcion de perdida y el gradiente en funcion de theta:
    f = lambda theta: 1/2*anp.linalg.norm(y - h(theta, X), 2)**2
    Df = grad(f)

    #Inicializamos el vector de parametros
    theta=np.zeros(X.shape[1])

    #optimizamos los parametros
    theta = optimizer(f, Df, theta, alfa=0.1, itmax=itmax)
    
    #Ya con los parametros optimizados definimos la funcion regresora con los parametros fijos
    regressor = lambda x: h(theta, x)
    regressor.params = theta

    #La funcion devuelve la funcion regresor para hacer predicciones
    return regressor
```

```python
X=np.array([[1],
   [2],
   [3]])
y=np.array([1, 2, 3])
regressor = linear_regression(X, y)
print(regressor(np.array([4]))) #Output: 4.000167772159999
print(regressor.params) #Output: [1.00004194]
```

```python
X=np.array([[1],
   [2],
   [3],
   [4]])
y=np.array([1, 1, 0, 0])
regressor = linear_regression(X, y)
print(regressor(np.array([5]))) #Output: 0.19187911993329507
print(regressor.params) #Output: [0.28756921]
```

En este ejemplo lo que esta pasando es que la funcion de perdida tiene un pozo profundo con gradientes muy grandes en modulo y nuestro alfa es muuy grande por lo que el descenso de gradiente diverge del minimo oscilando de un lado a otro. Aqui vemos que aunque la funcion sea convexa y con buenas propiedades puede pasar que el algortimo diverja. Podemos probar con algunos optimizadores del capitulo anterior o implementar el metodo de Newton especifico para funciones convexas

Para seguir minimizando en vez de maximizar para regresion logistica usaremos la funcion de perdida y el gradiente:
$$
\mathcal{L}(\theta) = - \ell(\theta) = - \sum_{i=1}^{n} \Big( y_i \log h_\theta(x_i) + (1 - y_i) \log (1 - h_\theta(x_i)) \Big)
$$
$$
\nabla_\theta \mathcal{L}(\theta) = \sum_{i=1}^{n} x_i \, \big( h_\theta(x_i) - y_i \big)
$$

```python
def logistic_regression(X, y, itmax=200, optimizer=gradient_descent):
    '''
    Para este caso tenemos que hay que estimar 1 parametro por variable a partir de un dataset con n ejemplos con m variables cada uno. 
    Por lo tanto las dimensiones son:
    X (Dataset):  nxm
    y (Labels): Como entrada 1xn
    theta (Parametros): 1xm
    '''
    #Definimos la funcion hipotesis
    h = lambda theta, x: 1/(1+anp.exp(x@theta))
    
    #Definimos la funcion de perdida y el gradiente en funcion de theta:
    f = lambda theta: -anp.sum(y*anp.log(h(theta, X)) + (1-y)*anp.log(1-h(theta, X)))
    Df = grad(f)

    #Inicializamos el vector de parametros
    theta=np.zeros(X.shape[1])

    #Optimizamos los parametros
    theta = optimizer(f, Df, theta, alfa=0.1, itmax=itmax)

    #Ya con los parametros optimizados definimos la funcion regresora y añadimos el valor de los parametros como argumento
    regressor = lambda x: h(theta, x)
    regressor.params = theta

    #La funcion devuelve la funcion regresor para hacer predicciones
    return regressor
```

```python
X=np.array([[1],
   [2],
   [3],
   [4]])
y=np.array([1, 1, 0, 0])
regressor = logistic_regression(X, y)
print(regressor(np.array([5]))) #Output: 0.19187911993329507
print(regressor.params) #Output: [0.28756921]
```

## Soft Max y **Multi-Class Classification**
Veamos otro ejemplo, este un poco mas elaborado. Supongamos que ahora nuestros datos no se distribuyen de forma continua ni es dos clases si no que lo hacen en un numero discreto superior a dos de clases. Pueden ser de tipo 1, 2, 3, ..., k. En este caso supongamos que nuestros datos se distribuyen siguiendo una distribucion multinomial de parametros $p_1, p_2, p_3, ..., p_k$. La funcion de densidad sera:
$$
f(x_1, x_2, \dots, x_k; n, p_1, \dots, p_k) =
P(X_1 = x_1, \dots, X_k = x_k) =
\frac{n!}{x_1! x_2! \cdots x_k!} \, p_1^{x_1} p_2^{x_2} \cdots p_k^{x_k},
$$
Ahora escribirla de forma que se pueda encuadrar como una distribucion de la familia exponencial resulta delicado. Llegamos a una expresion de la forma:
$$
f(x \mid \eta) 
= \frac{n!}{x_1! \cdots x_k!} \,
\exp\!\Bigg( \sum_{i=1}^{k} x_i \,\log p_i \Bigg),
$$
Aqui debemos darnos cuenta que para que la distribucion pertenezca a la familia exponenical el parametro natural $\eta$ debe ser libre y pertenecera los reales por lo tanto no lo podemos cojer como $\log p_i$. Debemos cogerlo como: $\eta_i = \log\!\left(\frac{p_i}{p_k}\right), \quad i=1,\dots,k-1$ donde $p_k$ es el parametro de referencia y el resto de parametros lo que miden es cuanto mas o menos probable es una categoria respecto a la de referencia. Haciendo esto el parametro natural ya no esta restringido y la distribucion pertenece a la familia exponencial haciendo:  

$T(y) = (y_1, y_2, ..., y_{k-1})$  
$\eta = \Bigg(\log\!\left(\frac{p_1}{p_k}\right), \log\!\left(\frac{p_2}{p_k}\right), ..., \log\!\left(\frac{p_{k-1}}{p_k}\right)\Bigg)$  
$a(\eta)= n \, \log\!\left( 1 + \sum_{i=1}^{k-1} e^{\eta_i} \right)$     

De aqui sabemos que por propiedades de la exponencial:
$$
\mathbb{E}[Y=i\mid X, n]=np_i=\frac{\partial A(\eta)}{\partial \eta_i}=n \, \frac{e^{\eta_i}}{1 + \sum_{j=1}^{k-1} e^{\eta_j}}=n \, \frac{e^{\eta_i}}{\sum_{j=1}^{k} e^{\eta_j}}
$$
La ultima igualdad se justifica imponiendo $\eta_k=1$ que es coherente con el desarrollo previo. Ahora para el caso n=1 y definiendo el enlace tipico de un modelo lineal $\eta=\theta x$ tenemos nuestra funcion de hipotesis:
$$
\mathbb{E}[Y=i\mid X]=\frac{e^{\theta_ix}}{\sum_{j=1}^{k} e^{\theta_jx}} := h_{\theta_i}(x)
$$
**Esta funcion recibe el nombre de softmax** y sera nuestra funcion de hipotesis, falta elegir los parametros optimos.  
Como hemos hecho antes definamos la funcion de verosimilitud:
$$
L(\theta) = \prod_{m=1}^{n} P\big(y^{(m)} \mid X^{(m)}; \theta\big) = \prod_{m=1}^{n} h_\theta\big(X^{(m)}\big)_{y^{(m)}}
$$
Esto es el producto de las probabilidades que nuestro modelo da a la categoria correcta. Aqui introduciremos la notacion de **one-hot encoding**, que es: en vez de representar $y^{(m)}$ por un entero lo haremos mediante un vector que tiene 0 en todas las posiciones menos en la correspondiente la valor de $y^{(m)}$ donde hay un 1. Por ejemplo si k=3 y $y^{(m)}=2$ entonces hacemos $y^{(m)}=[0, 1, 0]$. De esta forma la verosimilitud queda como:
$$
L(\theta) = \prod_{m=1}^{n} y^{(m)}h_\theta\big(X^{(m)}\big)
$$
$$
\ell(\theta) = \sum_{m=1}^{n} \log \big( y^{(m)} \cdot h_\theta(X^{(m)}) \big)
$$
Entonces definimos la funcion de perdida como:
$$
\mathcal{L}(\theta) = - \ell(\theta) = -\sum_{m=1}^{n} \log \big( y^{(m)} \cdot h_\theta(X^{(m)}) \big)
$$
con gradiente: 
$$
\nabla_\theta \mathcal{L}(\theta) = - \sum_{m=1}^n \big( y^{(m)} - h_\theta(X^{(m)}) \big) \, (X^{(m)})^\top
$$
Esta funcion de perdida se llama **cross entropy**

```python
def multiclass_regressor(X, y, k, itmax=200, optimizer=gradient_descent):
    '''
    Para este caso tenemos que hay que estimar k parametros por variable a partir de un dataset con n ejemplos con m variables cada uno. 
    Por lo tanto las dimensiones son:
    X (Dataset):  nxm
    y (Labels): Como entrada 1xn al pasar a one-hot nxk
    theta (Parametros): mxk (almacena como afecta cada variable a cada parametro)
    '''
    #Definimos la funcion de hipotesis softmax
    h = lambda theta, x: anp.exp(x@theta)/anp.sum(anp.exp(x@theta), axis=1, keepdims=True) #devuelve nxk

    #Transformamos y a one-hot encoding
    y_one_hot = np.eye(k)[y]

    #Definimos la funcion de perdida y el gradiente
    epsilon = 1e-12 #sumamos epsilon antes de aplicar log para evitar log(0)
    f = lambda theta: -anp.sum(anp.log(anp.sum(y_one_hot*h(theta, X), axis=1) + epsilon))
    Df = grad(f)

    #Inicializamos el vector de parametros
    theta = np.zeros((X.shape[1], k))

    #Optimizamos los parametros
    theta = optimizer(f, Df, theta, itmax=itmax)

    #Ya con los parametros optimizados definimos la funcion regresora y añadimos el valor de los parametros como argumento
    regressor = lambda x: h(theta, x)
    regressor.params = theta

    #La funcion devuelve la funcion regresor para hacer predicciones
    return regressor
```

```python
X = np.array([[1, 2],
              [2, 1],
              [3, 3]])

y = np.array([0, 2, 1])
k = 3
regressor = multiclass_regressor(X, y, k, itmax=1000, optimizer=newton)
predictions = regressor(np.array([[1, 1]])); #Output: [[0.05270569 0.89458861 0.05270569]]
```
